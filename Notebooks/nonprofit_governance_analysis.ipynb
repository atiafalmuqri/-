```python
classifier = LegalTopicClassifier()
df_classified = classifier.classify_dataframe(df_legal)

# Display top classifications with confidence score and evidence keywords
df_classified[['article_num', 'page_num', 'article_title', 'ai_predicted_topic', 'ai_confidence_score', 'ai_evidence_keywords']].head(12)
```
